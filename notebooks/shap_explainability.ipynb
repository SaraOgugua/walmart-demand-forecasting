
## Interpretation and Limitations

The local SHAP explanation identifies the 28-day rolling average as the largest contributor to the selected prediction (SHAP value = +0.374), followed by the 7-day rolling average (+0.093). The weekday feature contributed negatively (-0.034). These values explain the model output for this observation; they do not establish causal relationships.

The analysis used a permutation explainer after the TreeExplainer additivity check failed. The explanation sample was small, so the results should be treated as diagnostic rather than conclusive.

Model explanations must be considered alongside predictive performance. The selected model improved overall test MAE compared with the baseline but performed worse for low-demand products. Their high frequency of zero-sales observations indicates that intermittent demand remains a material limitation. Further validation and segment-specific improvements are required before operational use.
